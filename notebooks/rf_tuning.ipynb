```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from src.data.preprocess import notebook_root
from src.models.data import load_baseline_inputs
import src.models.random_forest as rf

root = notebook_root()
inputs = load_baseline_inputs(root)
```

```python
import joblib

results_dir = root / "results" / "random_forest"
results_dir.mkdir(parents=True, exist_ok=True)

tuning_results.to_csv(results_dir / "tuning_results.csv", index=False)

for feature_set, model in best_models.items():
    joblib.dump(model, results_dir / f"{feature_set}_tuned.joblib")
```
