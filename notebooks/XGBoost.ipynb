```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from src.data.preprocess import notebook_root
from src.models.data import load_baseline_inputs

import src.models.xgboost_model as xgb_model

root = notebook_root()
inputs = load_baseline_inputs(root)

print("Data loaded successfully.")
```

```python
models = xgb_model.fit_xgboost_models(inputs)

print("XGBoost models trained successfully.")
```

```python
results = xgb_model.evaluate_xgboost_models(inputs, models)

for feature_set, result in results.items():
    print(f"\n{feature_set.upper()}")

    print("\nMetrics:")
    display(result["metrics"])

    print("\nPer-Class Metrics:")
    display(result["per_class"])

    print("\nConfusion Matrix:")
    display(result["confusion"])
```

```python
import pandas as pd

metrics_df = pd.DataFrame([
    result["metrics"]
    for result in results.values()
])

display(metrics_df)

best_model = metrics_df.sort_values(
    "macro_f1",
    ascending=False
).iloc[0]

print("\nBest XGBoost model:")
print("Feature set:", best_model["configuration"])
print("Macro-F1:", round(best_model["macro_f1"], 4))
print("Accuracy:", round(best_model["accuracy"], 4))
```
